## 2 · How much differs, and where is the correspondence uncertain?
**Read the plot:** each narrow column is one alignment column, including inserted gaps. Teal = same amino acid; orange = different amino acids; pink = a gap. Purple below the strip = mapping changes among tested equally optimal alignments. Alignment columns are **not** UniProt residue numbers.

The representative alignment is the first optimum returned by the pinned software. All optimal alignments under four settings contribute to stability flags; stable means stable **within these tests**, not proven biological homology at that residue.


````python
# @title Recompute alignment, all tested optima, and bidirectional maps
matrix=substitution_matrices.load('BLOSUM62')
settings=[(-10,-.5),(-8,-.5),(-12,-.5),(-10,-1)]
def make_aligner(op,ex):
 a=Align.PairwiseAligner(); a.mode='global'; a.substitution_matrix=matrix; a.open_gap_score=op; a.extend_gap_score=ex; return a
def rows_for(a):
 hs,ps=str(a[0]),str(a[1]); hi=pi=0; rows=[]
 for col,(h,p) in enumerate(zip(hs,ps),1):
  hi+=h!='-'; pi+=p!='-'
  rows.append({'alignment_column':col,'human_position':hi if h!='-' else None,'human_aa':h,'pig_position':pi if p!='-' else None,'pig_aa':p,'kind':'gap' if '-' in (h,p) else 'match' if h==p else 'substitution','blosum62':float(matrix[h,p]) if '-' not in (h,p) else np.nan})
 return rows
hsets={i:set() for i in range(1,len(human)+1)}; psets={i:set() for i in range(1,len(pig)+1)}
hbase={i:set() for i in hsets}; pbase={i:set() for i in psets}; sensitivity=[]
for si,(op,ex) in enumerate(settings):
 aligns=make_aligner(op,ex).align(human,pig); n=len(aligns)
 assert n<10000,'Too many optima: review rather than silently truncate'
 sensitivity.append({'gap_open':op,'gap_extend':ex,'optimal_score':aligns.score,'optimal_alignments':n})
 if si==0: chosen=aligns[0]; base_count=n
 for a in aligns:
  for r in rows_for(a):
   h,p=r['human_position'],r['pig_position']
   if h is not None: hsets[h].add(p); (hbase[h].add(p) if si==0 else None)
   if p is not None: psets[p].add(h); (pbase[p].add(h) if si==0 else None)
df=pd.DataFrame(rows_for(chosen))
for c in ['human_position','pig_position']: df[c]=df[c].astype('Int64')
def stable(r,hm,pm):
 return (pd.isna(r.human_position) or len(hm[int(r.human_position)])==1) and (pd.isna(r.pig_position) or len(pm[int(r.pig_position)])==1)
df['unique_across_baseline_optima']=df.apply(lambda r:stable(r,hbase,pbase),axis=1)
df['stable_across_tested_parameters']=df.apply(lambda r:stable(r,hsets,psets),axis=1)
assert ''.join(df.human_aa).replace('-','')==human and ''.join(df.pig_aa).replace('-','')==pig
assert sorted(df.human_position.dropna().tolist())==list(range(1,512))
assert sorted(df.pig_position.dropna().tolist())==list(range(1,511))
hmap={int(r.human_position):int(r.pig_position) for r in df.itertuples() if pd.notna(r.human_position) and pd.notna(r.pig_position)}
pmap={p:h for h,p in hmap.items()}; assert all(pmap[p]==h for h,p in hmap.items())
old=pd.read_csv(INPUT/'Human_Pig_Residue_Correspondence.csv')
for c in ['human_aa','pig_aa','kind','stable_across_tested_parameters']: assert df[c].tolist()==old[c].tolist(),c
counts=df.kind.value_counts(); N=len(df); paired=int(counts['match']+counts['substitution'])
for name,frame in [('Human_Pig_Residue_Correspondence',df),('human_to_pig',df[df.human_position.notna()]),('pig_to_human',df[df.pig_position.notna()]),('all_differences',df[df.kind!='match']),('alignment_sensitivity',pd.DataFrame(sensitivity))]: frame.to_csv(TABLE/(name+'.csv'),index=False)
(INPUT/'alignment.fasta').write_text('>human_Q9NZK5\n'+str(chosen[0])+'\n>pig_P58780\n'+str(chosen[1])+'\n')
fig,axs=plt.subplots(2,1,figsize=(12,4.3),gridspec_kw={'height_ratios':[1,1.5]})
colors=[TEAL,ORANGE,RED]; vals=df.kind.map({'match':0,'substitution':1,'gap':2}).to_numpy()
axs[0].imshow(vals[None,:],aspect='auto',interpolation='nearest',cmap=ListedColormap(colors),vmin=0,vmax=2,extent=[.5,N+.5,.6,1])
unc=df.loc[~df.stable_across_tested_parameters,'alignment_column']; axs[0].scatter(unc,np.full(len(unc),.3),c=PURPLE,marker='D',s=18)
axs[0].set(ylim=(0,1.3),yticks=[],xlabel='Alignment column (1-based; includes gaps)',title='Full-length human–pig ADA2 comparison')
axs[0].legend(handles=[Patch(color=c,label=l) for c,l in zip(colors+[PURPLE],['Match','Substitution','Gap','Mapping uncertainty'])],loc='upper center',bbox_to_anchor=(.5,1.6),ncol=4,frameon=False)
for y,(k,c) in enumerate(zip(['match','substitution','gap'],colors)):
 axs[1].barh(y,counts[k],color=c,height=.55); axs[1].text(counts[k]+4,y,str(counts[k]),va='center',weight='bold')
axs[1].set(yticks=[0,1,2],yticklabels=['Match','Substitution','Gap columns'],xlim=(0,400),xlabel='Number of alignment columns'); axs[1].invert_yaxis(); fig.tight_layout(); savefig(fig,'01_alignment')
note(f'{counts["match"]} matches, {counts["substitution"]} substitutions and {counts["gap"]} gap columns across {N} columns. Identity is {counts["match"]/N:.2%} with gaps included, or {counts["match"]/paired:.2%} over {paired} paired residues. These describe sequence correspondence, not functional equivalence.','')
print('Uncertain human positions:',[h for h,v in hsets.items() if len(v)>1]); table(pd.DataFrame(sensitivity))
````

## 3 · What chemical properties change?
**Rules, fixed in this notebook:** D/E = acidic; K/R = basic; H = titratable histidine; all others = neutral for this coarse classification. “Charge reversal” means D/E ↔ K/R. Histidine is separate. These labels are **not calculated electrostatic potential or measured charge at a given pH**.

Hydrophobic-category flag uses A,V,I,L,M,F,W,Y versus other residues. Proline and glycine gains/losses are separate structural-interest flags. BLOSUM62 < 0 is a substitution-matrix flag, not pathogenicity. Counts can overlap; the complete 149-row table preserves every change.


````python
# @title Classify all substitutions with transparent rules
subs=df[df.kind=='substitution'].copy()
def charge(a): return 'acidic' if a in 'DE' else 'basic' if a in 'KR' else 'histidine' if a=='H' else 'neutral'
subs['human_class']=subs.human_aa.map(charge); subs['pig_class']=subs.pig_aa.map(charge)
subs['class_change']=subs.human_class!=subs.pig_class
subs['charge_reversal']=subs.apply(lambda r:{r.human_class,r.pig_class}=={'acidic','basic'},axis=1)
subs['hydrophobic_category_change']=subs.human_aa.isin(list('AVILMFWY'))!=subs.pig_aa.isin(list('AVILMFWY'))
subs['proline_change']=(subs.human_aa=='P')|(subs.pig_aa=='P'); subs['glycine_change']=(subs.human_aa=='G')|(subs.pig_aa=='G')
subs['negative_BLOSUM62']=subs.blosum62<0
flags=['class_change','charge_reversal','hydrophobic_category_change','proline_change','glycine_change','negative_BLOSUM62']
flag_counts=subs[flags].sum().astype(int)
fig,axs=plt.subplots(1,2,figsize=(12,4.4),gridspec_kw={'width_ratios':[1.3,1]})
labels=['Chemical class changes','Acidic ↔ basic','Hydrophobic category changes','Proline gained / lost','Glycine gained / lost','BLOSUM62 < 0']
axs[0].barh(labels,flag_counts,color=[TEAL,RED,ORANGE,PURPLE,'#4D8DAB',GRAY]); axs[0].invert_yaxis(); axs[0].set(xlabel='Substitutions (overlapping flags)',title='Properties of the 149 substitutions')
for y,n in enumerate(flag_counts): axs[0].text(n+.8,y,str(n),va='center')
axs[0].set_xlim(0,max(flag_counts)*1.22)
classes=['acidic','basic','histidine','neutral']; ct=pd.crosstab(subs.human_class,subs.pig_class).reindex(index=classes,columns=classes,fill_value=0)
im=axs[1].imshow(ct,cmap='YlGnBu'); axs[1].set(xticks=range(4),xticklabels=classes,yticks=range(4),yticklabels=classes,xlabel='Pig class',ylabel='Human class',title='Which class changed to which?')
for i in range(4):
 for j in range(4): axs[1].text(j,i,str(ct.iloc[i,j]),ha='center',va='center',color='white' if ct.iloc[i,j]>ct.values.max()/2 else INK)
fig.tight_layout(); savefig(fig,'02_properties')
note(f'{flag_counts["charge_reversal"]} substitutions reverse the acidic/basic category; {flag_counts["class_change"]} change the coarse chemical class. This identifies interpretable differences to inspect, but does not measure their effect.','')
table(subs[subs.charge_reversal][['human_position','human_aa','pig_position','pig_aa','stable_across_tested_parameters']])
````

## 4 · Conserved functional sites versus changed annotated sites
Join the alignment to the **frozen human UniProt annotations**, retaining their evidence codes and publication IDs. A single-residue site, an interval and a disulfide bond have different meanings: a disulfide annotation applies to its **two endpoints**, not every residue between them. An annotated binding interval does not establish direct ligand contact at every position.

**Evidence:** human glycosylation annotations carry experimental evidence (ECO:0000269). Active-site annotations here carry curator inference (ECO:0000305); some binding/region entries do not supply a feature-level evidence code. Preserve that distinction instead of labeling all sites experimentally validated.

## 5 · Put sequence numbers into experimental structural coordinates
Rebuild the mapping directly from **SIFTS XML**, then verify amino-acid identities and coordinate presence in **mmCIF atom records**. Track human UniProt position, PDB author residue ID, insertion code, chain and mmCIF label numbering separately.

**Sources:** experimental human ADA2 structures [3LGD](https://www.rcsb.org/structure/3LGD) and [3LGG](https://www.rcsb.org/structure/3LGG); [PDBe SIFTS](https://www.ebi.ac.uk/pdbe/docs/sifts/). Both chains are examined. No pig coordinates are generated.


````python
# @title Rebuild and check sequence-to-structure mapping
import xml.etree.ElementTree as ET
ns={'s':'http://www.ebi.ac.uk/pdbe/docs/sifts/eFamily.xsd'}
structure_rows=[]; atom_tables={}
for sid in ['3lgd','3lgg']:
 cif=MMCIF2Dict(str(INPUT/(sid+'.cif')))
 keys=['group_PDB','label_asym_id','auth_asym_id','auth_seq_id','label_seq_id','pdbx_PDB_ins_code','label_comp_id','type_symbol','label_alt_id','pdbx_PDB_model_num','Cartn_x','Cartn_y','Cartn_z','label_atom_id']
 atoms=pd.DataFrame({k:cif['_atom_site.'+k] for k in keys}); atoms=atoms[atoms.pdbx_PDB_model_num=='1'].copy()
 for c in ['Cartn_x','Cartn_y','Cartn_z']: atoms[c]=atoms[c].astype(float)
 atoms['author_key']=atoms.auth_seq_id+atoms.pdbx_PDB_ins_code.replace({'?':'','.':''})
 atom_tables[sid]=atoms
 root=ET.fromstring(gzip.decompress((INPUT/(sid+'_sifts.xml.gz')).read_bytes()))
 for r in root.findall('.//s:residue',ns):
  refs=r.findall('s:crossRefDb',ns); u=next((x for x in refs if x.get('dbSource')=='UniProt' and x.get('dbAccessionId')=='Q9NZK5'),None); p=next((x for x in refs if x.get('dbSource')=='PDB'),None)
  if u is None or p is None or p.get('dbChainId') not in ['A','B']: continue
  hp=int(u.get('dbResNum')); chain=p.get('dbChainId'); aid=p.get('dbResNum'); aa=u.get('dbResName'); assert human[hp-1]==aa
  observed=atoms[(atoms.auth_asym_id==chain)&(atoms.author_key==aid)&(atoms.group_PDB=='ATOM')]
  if len(observed): assert set(observed.label_comp_id.map(seq1))=={aa}
  not_observed=any(x.text=='Not_Observed' for x in r.findall('s:residueDetail',ns))
  assert bool(len(observed)) != not_observed, (sid,chain,hp,'SIFTS/coordinate disagreement')
  d=df[df.human_position==hp].iloc[0]
  structure_rows.append({'structure':sid.upper(),'chain':chain,'human_position':hp,'human_aa':aa,'pig_position':d.pig_position,'pig_aa':d.pig_aa,'kind':d.kind,'stable_across_tested_parameters':d.stable_across_tested_parameters,'author_residue_id':aid,'pdbe_residue_number':r.get('dbResNum'),'label_asym_id':observed.label_asym_id.iloc[0] if len(observed) else None,'label_seq_id':observed.label_seq_id.iloc[0] if len(observed) else None,'insertion_code':observed.pdbx_PDB_ins_code.iloc[0] if len(observed) else None,'coordinate_present':bool(len(observed)),'mapping_source':'SIFTS XML + mmCIF atom verification'})
sm=pd.DataFrame(structure_rows); assert not sm.duplicated(['structure','chain','human_position']).any()
sm.to_csv(TABLE/'Sequence_to_Structure_Residue_Mapping.csv',index=False)
coverage=[]; full=[]
for sid in ['3LGD','3LGG']:
 for ch in ['A','B']:
  part=sm[(sm.structure==sid)&(sm.chain==ch)].set_index('human_position'); states=[]
  for hp in range(1,512):
   status=0 if hp not in part.index else 2 if part.loc[hp,'coordinate_present'] else 1; states.append(status)
   full.append({'structure':sid,'chain':ch,'human_position':hp,'status':['outside_SIFTS','mapped_without_coordinates','mapped_with_coordinates'][status]})
  coverage.append(states)
full=pd.DataFrame(full); full.to_csv(TABLE/'structure_coverage_all_positions.csv',index=False)
fig,ax=plt.subplots(figsize=(12,3)); ax.imshow(coverage,aspect='auto',interpolation='nearest',cmap=ListedColormap([GRAY,ORANGE,TEAL]),vmin=0,vmax=2,extent=[.5,511.5,3.5,-.5])
ax.hlines([.5,1.5,2.5],.5,511.5,color='white',lw=9)
ax.set(yticks=range(4),yticklabels=['3LGD A','3LGD B','3LGG A','3LGG B'],xlabel='Human UniProt Q9NZK5 position',title='Experimental coordinate coverage · exact positions')
ax.legend(handles=[Patch(color=c,label=l) for c,l in zip([GRAY,ORANGE,TEAL],['Outside SIFTS','Mapped, no coordinates','Coordinates verified'])],loc='upper center',bbox_to_anchor=(.5,1.4),ncol=3,frameon=False); savefig(fig,'04_structure_coverage')
covtable=full.groupby(['structure','chain','status']).size().unstack(fill_value=0).reset_index(); table(covtable)
note('Sequence-to-structure correspondence and coordinate availability are separate checks. Only residues with verified atoms enter the geometry analysis.','')
````

## 6 · Structural context, with a defined distance calculation
For every stable substitution with coordinates, compute the **minimum heavy-atom distance** to annotated catalytic/zinc-binding residues in the **same chain**. Exclude H/D atoms; retain blank/A alternate locations; use model 1 of each structure. Report each chain separately, plus the range across four chain/structure combinations.

“Near” = ≤5 Å; sensitivity flags also use 4 and 6 Å. These are descriptive neighborhoods, **not a functional-effect threshold**. A site's distance to a catalytic residue is not its distance to substrate. Interchain interfaces, solvent exposure, dynamics and pig-specific geometry are not assessed.


````python
# @title Compute human-structure neighborhoods for all stable substitutions
anchors=sorted(set(annotated.loc[(annotated.type=='Active site')|((annotated.type=='Binding site')&(annotated.description=='Zn(2+)')),'human_position'].astype(int)))
dist_rows=[]
for sid in ['3lgd','3lgg']:
 for ch in ['A','B']:
  amap=sm[(sm.structure==sid.upper())&(sm.chain==ch)&sm.coordinate_present].set_index('human_position')
  atoms=atom_tables[sid]; atoms=atoms[(atoms.auth_asym_id==ch)&(atoms.group_PDB=='ATOM')&(~atoms.type_symbol.isin(['H','D']))&atoms.label_alt_id.isin(['.','?','A'])]
  def coords(h): return atoms.loc[atoms.author_key==amap.loc[h,'author_residue_id'],['Cartn_x','Cartn_y','Cartn_z']].to_numpy()
  assert all(h in amap.index for h in anchors)
  for r in subs[subs.stable_across_tested_parameters].itertuples():
   hp=int(r.human_position)
   if hp not in amap.index: continue
   xyz=coords(hp); ds=[float(np.linalg.norm(xyz[:,None,:]-coords(a)[None,:,:],axis=2).min()) for a in anchors]
   j=int(np.argmin(ds)); dist_rows.append({'human_position':hp,'structure':sid.upper(),'chain':ch,'distance_A':ds[j],'nearest_anchor_human':anchors[j]})
dist=pd.DataFrame(dist_rows); dist.to_csv(TABLE/'catalytic_neighborhood_distances_all_chains.csv',index=False)
dsummary=dist.groupby('human_position').agg(min_distance_A=('distance_A','min'),max_distance_A=('distance_A','max'),mean_distance_A=('distance_A','mean'),n_chain_structures=('distance_A','size')).reset_index()
for cut in [4,5,6]: dsummary[f'within_{cut}A_in_any']=dsummary.min_distance_A<=cut; dsummary[f'within_{cut}A_in_all']=dsummary.max_distance_A<=cut
subs=subs.merge(dsummary,on='human_position',how='left')
fig,ax=plt.subplots(figsize=(12,4)); plot=subs[subs.min_distance_A.notna()]
ax.vlines(plot.human_position,plot.min_distance_A,plot.max_distance_A,color=GRAY,lw=2)
ax.scatter(plot.human_position,plot.mean_distance_A,s=20,c=np.where(plot.charge_reversal,RED,TEAL),zorder=3)
ax.axhline(5,color=ORANGE,ls='--',label='5 Å descriptive cutoff'); ax.set(xlabel='Human UniProt position (stable substitutions only)',ylabel='Distance to nearest anchor (Å)',title='Distance to catalytic / zinc-binding residues in human ADA2'); ax.legend(handles=[plt.Line2D([],[],marker='o',ls='',color=TEAL,label='Other stable substitutions'),plt.Line2D([],[],marker='o',ls='',color=RED,label='Acidic/basic reversal'),plt.Line2D([],[],color=GRAY,label='Range across 4 chains/structures'),plt.Line2D([],[],color=ORANGE,ls='--',label='5 Å descriptive cutoff')],frameon=False,fontsize=8,ncol=2); savefig(fig,'05_structure_neighborhoods')
note(f'{len(plot)} stable substitutions have coordinates; {int(dsummary.within_5A_in_any.sum())} lie within 5 Å in at least one tested chain/structure, and {int(dsummary.within_5A_in_all.sum())} do so in all four. Gray ranges show variation between structures/chains; red points are acidic/basic reversals.','')
````
